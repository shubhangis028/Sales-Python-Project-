```python
# Drop null values or inconsistancy
df=df.dropna()
df
```

```python
df.shape
```

```python
df.head()
```

```python
# Set index for table
df.set_index('Row ID')
```

```python
# Rename Columns
df.rename(columns={'Country/Region':'Country'})
```

```python
df.describe()
```

```python
# list the top products Category
c=df.groupby('Category')['Profit'].sum().sort_values(ascending = False)
c
```

```python
plt.figure(figsize=(6,4))
plt.xlabel('Total Sales')
plt.ylabel('Category')
plt.title('Top products Category')
r=["darkcyan","lightgreen","orange"]
plt.bar(c.index,c,color=r,width=0.4, alpha=1)
plt.show()
```

```python
# list highest sales in region
d=df.groupby('Region')['Sales'].sum().sort_values(ascending= False).head(5)
d
```

```python
plt.figure(figsize=(6, 4))
plt.xlabel('Total Sales')
plt.ylabel('Region')
plt.title('Top products Category')
o=["darkcyan","lightgreen","orange","lightblue"]
plt.barh(d.index, d, color=o, height=0.6)
plt.show()
```

```python
# Segments breakdown
e=df['Segment'].value_counts()
e
```

```python
plt.figure(figsize=(4,4))
plt.pie(e, labels=e.index,autopct='%.2f%%')
plt.title('Segment breakdown')
plt.show()
```

```python
# highest sales by year
df.loc[:,'Year']=df['Order Date'].dt.year
df['Year']
```

```python
Y=df.groupby('Year')['Sales'].sum().sort_values(ascending=False)
Y
```

```python
plt.figure(figsize=(6,4))
plt.plot(Y.index,Y,"brown",marker='o',markersize=8,linestyle="-.")
plt.xlabel('Highest sales by year')
plt.ylabel('Total Sales')
plt.title('Year')
plt.show()
```

```python
# Average delivery time
df.loc[:,"Delevery Days"]=(df['Ship Date']-df['Order Date']).dt.days
df["Delevery Days"].mean()
```

```python
# distribution of discounts
plt.figure(figsize=(6,4))
df['Discount'].plot(kind='hist', bins=10, color='darkcyan', title='Discount Distribution')
plt.xlabel('Discount %')
plt.show()
```

```python
# 10 states have the highest sales
plt.figure(figsize=(6,4))
df.groupby('State')['Sales'].sum().sort_values(ascending= False).head(10).plot(kind='barh',color= 'darkcyan',title="10 states have the highest sales")
plt.xlabel('Total Sales')
plt.show()
```

```python
# Relationship between discount and profit
plt.figure(figsize=(6,4))
df.plot.scatter(x='Discount', y='Profit',color='lightseagreen', marker='*', title='Discount vs. Profit')
plt.show()
# plt.grid()
# or
```

```python
sns.regplot(x='Discount', y='Profit', data=df)
plt.title('Relationship between Discount and Profit')
plt.show()
```

```python
# Monthly sales Trend
df.loc[:,'Month'] = df['Order Date'].dt.month
df.groupby('Month')['Sales'].sum().plot(kind='line', color='orange', marker='o', linestyle='-.',  title='Monthly Sales Trend')
plt.show()
```

```python
# Outlier products with very high or low unit price (Sales / Quantity)
df.loc[:,'Unit Price']=df['Sales']/df['Quantity']
sns.boxplot(data=df, x='Category', y='Unit Price')
plt.title('Unit Price Distribution by Category')
plt.show()
```
